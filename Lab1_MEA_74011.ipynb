## 📚 Wstęp teoretyczny — zanim zaczniemy

### Czym jest bezpieczeństwo systemów?

**Bezpieczeństwo systemu operacyjnego** to ochrona danych i zasobów komputera przed:

- 🔓 **nieautoryzowanym dostępem** (gdy ktoś czyta dane, do których nie ma prawa),
- ✏️ **nieautoryzowaną modyfikacją** (gdy ktoś zmienia dane bez zgody właściciela),
- 🚫 **zakłóceniem działania** (gdy system przestaje działać poprawnie).

### Trzy filary bezpieczeństwa — triada CIA 🔺

| Filar | Po angielsku | Pytanie, na które odpowiada |
|---|---|---|
| 🔒 **Poufność** | Confidentiality | *Kto może zobaczyć dane?* |
| ✔️ **Integralność** | Integrity | *Czy dane nie zostały zmienione?* |
| 🟢 **Dostępność** | Availability | *Czy dane są dostępne, gdy są potrzebne?* |

### Najważniejsze pojęcia na dziś

- **System operacyjny (OS)** — program zarządzający całym komputerem: pamięcią, dyskiem, plikami i procesami. Przykłady: Windows, Linux, macOS.
- **Użytkownik** — każda osoba korzystająca z komputera ma w systemie swoją **tożsamość** (nazwę i numer identyfikacyjny).
- **Uprawnienia** — reguły mówiące, **kto** może **co** zrobić z danym zasobem (np. czytać plik, zmieniać go, uruchomić).
- **Proces** — uruchomiony program. Każdy proces należy do jakiegoś użytkownika.
- **Katalog domowy** — prywatny folder użytkownika (w Linuksie: `/home/nazwa_użytkownika`).

## 📁 Zadanie 3: Pliki i uprawnienia — kto może co?

> ### 📘 Wstęp teoretyczny
>
> Każdy plik w Linuksie ma **właściciela** i **uprawnienia**. Widzisz je po poleceniu `ls -l`, np.:
>
> `-rw-r--r-- 1 user user 1234 paź 4 10:00 raport.txt`
>
> Pierwsze 10 znaków to **maska uprawnień**, którą czytamy tak:
>
> | Znak | Znaczenie |
> |---|---|
> | `-` / `d` | zwykły plik / **d**irectory (katalog) |
> | `r`, `w`, `x` | prawa **r**ead (czytanie), **w**rite (pisanie), e**x**ecute (uruchamianie) |
>
> Trzy trójki po pierwszym znaku to prawa kolejno dla:
>
> 1. **właściciela** (owner),
> 2. **grupy** (group),
> 3. **innych** (others — wszyscy pozostali).
>
> Polecenie `chmod` zmienia uprawnienia, np.:
>
> - `chmod u+x plik` — dodaje (+) właścicielowi (u) prawo uruchamiania (x),
> - `chmod o-r plik` — odbiera innym (o) prawo czytania (r).
>
> 🔐 **Zasada bezpieczeństwa (zasada minimalnych uprawnień):** nikomu nie dawaj więcej uprawnień, niż naprawdę potrzebuje. Plik z danymi, które mają być poufne, **nie powinien** być czytelny dla "innych".

```python
# ZADANIE 3 — uprawnienia plików
!echo "Moja tajna notatka laboratoryjna" > tajne.txt
!ls -l tajne.txt

# >>> TU WPISZ KOD <<<
# 3.1 Odejmij prawo czytania "innym" (chmod o-r tajne.txt), potem wypisz ponownie ls -l tajne.txt:
!chmod o-r tajne.txt
!ls -l tajne.txt
# 3.2 Odejmij prawo czytania także grupie (chmod g-r tajne.txt). Kto teraz może przeczytać plik?
!chmod g-r tajne.txt
!ls -l tajne.txt
# Odpowiedź: Teraz tylko właściciel pliku (root) ma prawo odczytu.
# 3.3 Spróbuj odczytać plik:
!cat tajne.txt
#     Trop: sprawdź !whoami — w Colab jesteś rootem (administratorem),
#     a administrator może przeczytać plik NAWET bez prawa "r"!
#     Gdybyś był zwykłym użytkownikiem bez prawa "r", system odmówiłby odczytu
#     (Permission denied). Zapisz ten wniosek w komentarzu.
!chmod 640 tajne.txt
!ls -l tajne.txt

!rm tajne.txt
```

## 📂 Zadanie 4: Procesy — kto i co uruchamia programy

> ### 📘 Wstęp teoretyczny
>
> **Proces** to uruchomiony program. W każdej chwili system operacyjny utrzymuje listę procesów i pilnuje, by:
>
> - każdy proces należał do konkretnego **użytkownika** (procesy nie mogą "unikać" odpowiedzialności),
> - proces nie przekraczał swoich uprawnień (np. nie czytał cudzych plików).
>
> Najważniejsze polecenia:
>
> - `ps aux` — wypisuje listę wszystkich procesów (użytkownik, PID, polecenie),
> - `sleep 100` — uruchamia program, który "śpi" przez podaną liczbę sekund — idealny do eksperymentów,
> - `kill <PID>` — kończy proces o danym numerze PID (możesz kończyć tylko **własne** procesy!).
>
> 🔐 **Bezpieczeństwo:** granice między procesami to fundament bezpieczeństwa systemu. Wirus uruchomiony jako zwykły użytkownik nie może np. zakończyć procesów administratora ani czytać jego plików.

### ✅ Uruchom własny proces, znajdź go na liście i zakończ go.


```python
# ZADANIE 4 — procesy
import subprocess

# Uruchamiamy w tle program "sleep 300" (śpi przez 5 minut):
proc = subprocess.Popen(["sleep", "300"])
print("Uruchomiono proces sleep z numerem PID =", proc.pid)

# >>> TU WPISZ KOD <<<
# 4.1 Wypisz tylko linie dotyczące sleep (podpowiedź):
!ps aux | grep sleep
# 4.2 Sprawdź w wyniku, jaki UŻYTKOWNIK jest właścicielem procesu sleep. Zapisz w komentarzu:
#     Właścicielem procesu sleep jest: root
# 4.3 Zakończ proces poleceniem kill z jego PID (użyj numeru z 4.1 albo proc.pid):
!kill {proc.pid}
# 4.4 Sprawdź ponownie (!ps aux | grep sleep), czy proces zniknął.
!ps aux | grep sleep
```

## 🌍 Zadanie 5: Zmienne środowiskowe i pułapka PATH

> ### 📘 Wstęp teoretyczny
>
> **Zmienne środowiskowe** to ustawienia, które każdy proces dostaje od systemu przy starcie — np. `HOME` (katalog domowy), `USER` (nazwa użytkownika), a szczególnie ważna: **`PATH`**.
>
> `PATH` to lista katalogów, w których system **szuka programów**, gdy wpisujesz polecenie. Gdy napiszesz `ls`, system nie przeszukuje całego dysku — sprawdza po kolei katalogi z `PATH` i uruchamia **pierwszy znaleziony** plik o nazwie `ls`.
>
> ⚠️ **Atak przez PATH (dla dociekliwych):** gdyby ktoś umieścił złośliwy plik o nazwie `ls` w katalogu, który w PATH stoi **przed** systemowym, i nakłonił Cię do wpisania `ls` w tym katalogu — uruchomiłbyś jego program zamiast systemowego! Dlatego:
>
> - nigdy nie powinno się mieć `.` (bieżący katalog) w PATH,
> - przy wrażliwych operacjach używa się pełnych ścieżek, np. `/bin/ls`.

### ✅ Zbadaj swoje środowisko i przetestuj podstęp z PATH (bezpiecznie, na własnym programie).

## 🔑 Zadanie 6: Hasła — dlaczego nie przechowujemy haseł "na wprost"?

> ### 📘 Wstęp teoretyczny
>
> System **nigdy nie przechowuje Twojego hasła w postaci jawnej**. Zamiast tego zapisuje jego **skrót (hash)** — wynik funkcji haszującej, np. **SHA-256**. Skrót:
>
> - ma stałą długość (SHA-256 → zawsze 64 znaki),
> - jest **praktycznie nieodwracalny**: z hasha nie da się (rozsądnie) odtworzyć hasła,
> - zmienia się **całkowicie** przy najmniejszej zmianie wejścia (tzw. efekt lawiny).
>
> Gdy się logujesz, system haszuje to, co wpisałeś, i porównuje skróty — hasło może pozostać tajne nawet dla administratora.
>
> ⚠️ **Słabe strony:** proste hasła łatwo złamać metodą "słownika" — atakujący haszuje słowa ze słownika i porównuje. Dlatego dodaje się **sól (salt)** — losowy dodatek do hasła przed haszowaniem, który sprawia, że dwa identyczne hasła mają różne skróty — oraz używa **mocnych, długich haseł**.

### ✅ Policz skróty, zaobserwuj efekt lawiny i przekonaj się, że sól chroni przed słownikami.


```python
# ZADANIE 6 – hashowanie haseł
import hashlib

def sha256(text):
    return hashlib.sha256(text.encode()).hexdigest()

haslo_a = "haslo123"
haslo_b = "haslo124"     # różni się JEDNYM znakiem
haslo_c = "haslo123"     # identyczne z haslo_a

print("haslo_a ->", sha256(haslo_a))
print("haslo_b ->", sha256(haslo_b))
print("haslo_c ->", sha256(haslo_c))

# >>> TU WPISZ KOD <<<
# 6.1 Porównaj długości skrótów (podpowiedź: len(sha256("x"))).
#     Czy długość skrótu zależy od długości hasła?
print("Długość dla 'x':", len(sha256("x")))
print("Długość dla 'haslo123':", len(sha256("haslo123")))
# Odpowiedź: Długość skrótu ma zawsze 64 znaki. Nie zależy od długości hasła.

# 6.2 Czy skróty haslo_a i haslo_b są do siebie podobne? Opisz w komentarzu "efekt lawiny".
# Odpowiedź: Skróty są całkowicie inne. To jest efekt lawiny.

# 6.3 Dodaj sól: policz sha256("sol_abc" + haslo_a) oraz sha256("sol_xyz" + haslo_a).
print("Sól abc:", sha256("sol_abc" + haslo_a))
print("Sól xyz:", sha256("sol_xyz" + haslo_a))
#     Czy identyczne hasła z różną solą dają różne skróty? Odpowiedz w komentarzu.
# Odpowiedź: Tak, różna sól daje zupełnie inne skróty dla tego samego hasła.
```

```python
# ZADANIE 6 c.d. – mini atak słownikowy
import hashlib

# Udajemy skrót hasła "wykradziony" z systemu (oczywiście tylko do ćwiczenia!):
skrot_docelowy = hashlib.sha256("qwerty".encode()).hexdigest()
print("Skrót do złamania:", skrot_docelowy)

slownik = ["123456", "password", "qwerty", "admin", "iloveyou", "haslo123"]

# >>> TU WPISZ KOD <<<
# Dla każdego hasła ze słownika policz jego SHA-256 i porównaj ze skrot_docelowy.
# Jeśli znajdziesz dopasowanie – wypisz "Złamano hasło: ..." i przerwij pętlę (break).
# Skorzystaj z pętli for i funkcji hashlib.sha256(...).hexdigest().

for h in slownik:
    skrot = hashlib.sha256(h.encode()).hexdigest()
    if skrot == skrot_docelowy:
        print("Złamano hasło:", h)
        break
```

## 🧩 Zadanie 7: Integralność danych — sumy kontrolne

> ### 📘 Wstęp teoretyczny
>
> Skąd wiadomo, że plik **nie został zmieniony** — np. że pobrany instalator nie został podmieniony przez atakującego? Służy do temu **suma kontrolna (checksum)** — skrót pliku liczony tymi samymi funkcjami, które poznałeś w Zadaniu 6.
>
> Schemat pracy:
>
> 1. autor pliku publikuje plik **oraz** jego sumę kontrolną (np. SHA-256),
> 2. po pobraniu pliku **Ty** liczysz jego sumę kontrolną,
> 3. jeśli sumy się **zgadzają** → plik jest (prawie na pewno) nietknięty; jeśli się różnią → coś zostało zmienione.
>
> Polecenia w Linuksie: `sha256sum plik`, `md5sum plik`.
>
> ⚠️ Starsza funkcja **MD5 jest złamana** — da się celowo stworzyć dwa różne pliki o tym samym skrócie. Dlatego dziś zaleca się **SHA-256** lub nowsze.

### ✅ Zbuduj plik, sprawdź jego sumę, "przechwyć" go i wykryj modyfikację.


```python
# ZADANIE 7 – sumy kontrolne
!echo "Przelew: 100 zl na konto Boba" > umowa.txt
!sha256sum umowa.txt
!md5sum umowa.txt

# >>> TU WPISZ KOD <<<
# 7.1 Policz sumę ponownie (te same polecenia). Czy się zmieniła? Dlaczego (nie)?
!sha256sum umowa.txt
!md5sum umowa.txt
# Odpowiedź: Suma się nie zmieniła, bo plik jest taki sam.

# 7.2 Zmień JEDEN znak w pliku:
#     !echo "Przelew: 101 zl na konto Boba" > umowa.txt
#     a potem policz !sha256sum umowa.txt jeszcze raz.
!echo "Przelew: 101 zl na konto Boba" > umowa.txt
!sha256sum umowa.txt

# 7.3 Porównaj sumy sprzed i po zmianie. Czy podobne? Zapisz wniosek o integralności.
# Wniosek: Suma jest zupełnie inna. Zmiana jednego znaku niszczy integralność.

# 7.4 Porównaj długości sum MD5 i SHA-256. Która funkcja ma dłuższy skrót?
#     Dłuższy skrót = trudniej znaleźć "kolizję" (dwa pliki o tym samym skrócie).
# Odpowiedź: SHA-256 ma dłuższy skrót i jest bezpieczniejszy.

# 7.5 Posprzątaj: !rm umowa.txt
!rm umowa.txt
```

## 📦 Zadanie 8: Piaskownica — sprawdzamy izolację środowiska

> ### 📘 Wstęp teoretyczny

>
